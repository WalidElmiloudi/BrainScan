# Dataset location

```python
from pathlib import Path

root = Path.cwd().parents[0]
data_dir = root / "data" / "raw" / "BrainScan_Dataset"

print(f"the images are in : {data_dir}")
```

# Folder structure

```python
for folder in data_dir.iterdir():
    print(folder.name)
```

# Image counting

```python
for folder in data_dir.iterdir():
    image_count = sum(1 for item in folder.iterdir())
    print(f"there's {image_count} in {folder.name} directory")
```

# File validation

```python
FORMATS = [".jpg",".jpeg","png",".bmp"]
supported = []
unsupported = []
unreadable = []

for folder in data_dir.iterdir():
    for file_path in folder.rglob("*"):
        if file_path.is_file():
            try:
                with file_path.open("rb") as file :
                     _ = file.read(1024)
                if file_path.suffix.lower() in FORMATS :
                    supported.append(file_path)
                else :
                    unsupported.append(file_path)
            except (PermissionError , FileNotFoundError , OSError) as e :
                unreadable.append(file_path)

print(f"there's {len(supported)} supported images")
print(f"there's {len(unsupported)} unsupported images")
print(f"there's {len(unreadable)} unreadable images")
```
